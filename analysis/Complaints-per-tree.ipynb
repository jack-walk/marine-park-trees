```python
import pandas as pd
```

```python
# Importing tree count CSV file

tree_df = pd.read_csv("tree-count-2015.csv")
tree_df
```

```python
## Checking data types

tree_df["Zip code"].dtype
```

```python
tree_df["Tree count"].dtype
```

```python
## Converting Zip code from int to string, and converting Tree count to int

tree_df["Zip code"] = (
    tree_df["Zip code"]
    .astype(str)
    .str.replace(r"\.0$", "", regex=True)
)

tree_df["Tree count"] = pd.to_numeric(tree_df["Tree count"], errors="coerce").astype("Int64")
```

```python
tree_df
```

```python
# Import complaints CSV file

zip_df = pd.read_csv("zip_counts_updated.csv")

zip_df
```

```python
# Checking data types

zip_df["Incident Zip"].dtype
```

```python
zip_df["Count"].dtype
```

```python
# Converting Incident Zip to string

zip_df["Incident Zip"] = (
    zip_df["Incident Zip"]
    .astype(str)
    .str.replace(r"\.0$", "", regex=True)
)

zip_df
```

```python
# Merge
combined_df = pd.merge(
    zip_df,
    tree_df,
    left_on="Incident Zip",
    right_on="Zip code",
    how="inner"
)
```

```python
# New column
combined_df["Complaints per Tree"] = combined_df["Count"] / combined_df["Tree count"]
```

```python
# Test
combined_df
```

```python
combined_df.to_csv("combined_df.csv", index=False)
```

```python
combined_df = combined_df.sort_values(
    by="Complaints per Tree",
    ascending=False
).reset_index(drop=True)

combined_df
```
