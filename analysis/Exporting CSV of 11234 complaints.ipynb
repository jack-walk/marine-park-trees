Exporting individual complaints in zip code 11234 as a .CSV file to visualize on a locator map.

```python
import pandas as pd
```

```python
## Importing CSV file of all 311 service requests over tree overgrowth since 2020.
dfFull = pd.read_csv("311_updated.csv", low_memory=False)
```

```python
## Filtering out entries prior to 2023 for time relevance.
dfFull['Created Date'] = pd.to_datetime(dfFull['Created Date'], errors='coerce')
dfDateFiltered = dfFull[dfFull['Created Date'] >= pd.Timestamp('2023-01-01 00:00:00')]
```

```python
## Filtering out entries outside zip code 11234.
dfZipFiltered = dfDateFiltered[dfDateFiltered['Incident Zip'] == 11234.0]
```

```python
## Removing unnecessary decimals on zip code entries.
dfZipFiltered['Incident Zip'] = pd.to_numeric(dfZipFiltered['Incident Zip'], errors='coerce').astype('Int64')
dfZipFiltered
```

```python
## Exporting to .CSV file

dfZipFiltered.to_csv("Marine_Park_Complaints.csv", index=False)
```
